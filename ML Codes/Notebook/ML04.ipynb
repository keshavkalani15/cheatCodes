```python
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2.0
learning_rate = 0.1
max_iterations = 100
tolerance = 0.000001

x_values = [x]
y_values = [function(x)]
```

```python
for i in range(max_iterations):
    gradient = derivative(x)
    new_x = x - learning_rate * gradient

    x_values.append(new_x)
    y_values.append(function(new_x))

    if abs(new_x - x) < tolerance:
        x = new_x
        break

    x = new_x
```

```python
print("Iterations:", i + 1)
print("Minimum x:", round(x, 6))
print("Minimum y:", round(function(x), 6))
```

```python
x_curve = [value / 100 for value in range(-500, 101)]
y_curve = [function(value) for value in x_curve]

plt.plot(x_curve, y_curve, label="y = (x + 3)^2")
plt.plot(x_values, y_values, "ro-", label="Gradient Descent steps")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent")
plt.legend()
plt.grid(True)
plt.show()
```

## Practical 4 — Likely Viva Questions and Answers

1. **What is the aim of this practical?**  
   To implement Gradient Descent and use it to find the minimum of a function.

2. **What function did you use?**  
   \(y=(x+3)^2\), starting from \(x=2\).

3. **What is Gradient Descent?**  
   It is a method that repeatedly adjusts a value in the direction that reduces a function.

4. **What is the minimum of this function?**  
   The minimum is at \(x=-3\), where \(y=0\).

5. **What is a gradient?**  
   The gradient is the slope of the function at the current point.

6. **What is the derivative of \(y=(x+3)^2\)?**  
   \(\frac{dy}{dx}=2(x+3)\).

7. **What is the Gradient Descent update rule?**  
   \(x_{\text{new}}=x_{\text{old}}-\text{learning rate}\times\text{gradient}\).

8. **Why do we subtract the gradient?**  
   Subtracting moves \(x\) downhill, toward a lower function value.

9. **What is the learning rate?**  
   It controls the size of each update step.

10. **What happens if the learning rate is too small?**  
    The algorithm moves slowly and may need many iterations.

11. **What happens if the learning rate is too large?**  
    It may jump over the minimum or fail to settle near it.

12. **What is an iteration?**  
    One update of \(x\) using the current gradient.

13. **What is the purpose of the tolerance value?**  
    It lets the algorithm stop when updates become very small.

14. **Why set a maximum number of iterations?**  
    To ensure the loop stops even if it does not meet the tolerance condition.

15. **What does the plot show?**  
    The function’s curve and the sequence of points moving toward its minimum.

16. **Does Gradient Descent always find the global minimum?**  
    No. For more complex functions it may settle at a local minimum, depending on the starting point and function shape.

17. **What is the starting point in your code?**  
    \(x=2\), as specified in the assignment example.

18. **How can you check whether the algorithm worked?**  
    The printed \(x\) should be close to \(-3\), and \(y\) should be close to \(0\).
